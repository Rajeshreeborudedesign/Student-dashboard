# STUDENT RECORD MANAGEMENT SYSTEM

## Algorithm 1: Add Student

Step 1: Start.

Step 2: Read student details such as Roll Number,
Name, Department, Subjects, Marks, Attendance
and Contact Information.

Step 3: Check whether the Roll Number already exists.

Step 4: If the Roll Number exists, display
"Duplicate Roll Number" and stop the operation.

Step 5: Otherwise, create a new student record.

Step 6: Store the record in the student collection.

Step 7: Display "Student Added Successfully".

Step 8: Stop.

## Algorithm 2: Search Student

Step 1: Start.

Step 2: Read the Roll Number from the user.

Step 3: Search for the Roll Number in student records.

Step 4: If the Roll Number exists, display student details.

Step 5: Otherwise, display "Student Not Found".

Step 6: Stop.

## Algorithm 3: Update Student

Step 1: Start.

Step 2: Read the Roll Number from the user.

Step 3: Search for the Roll Number in student records.

Step 4: If the student is not found, display "Student Not Found".

Step 5: Otherwise, ask the user which field to update.

Step 6: Read the new value from the user.

Step 7: Update the selected field in the student record.

Step 8: Display "Student Updated Successfully".

Step 9: Stop.

## Algorithm 4: Delete Student

Step 1: Start.

Step 2: Read the Roll Number from the user.

Step 3: Search for the Roll Number in student records.

Step 4: If the student is not found, display "Student Not Found".

Step 5: Otherwise, delete the student record from the collection.

Step 6: Display "Student Deleted Successfully".

Step 7: Stop.

## Algorithm 5: Calculate Average

Step 1: Start.

Step 2: Read the Roll Number from the user.

Step 3: Search for the student in the records.

Step 4: If the student is not found, display "Student Not Found".

Step 5: Otherwise, read the student's marks.

Step 6: Calculate the total marks using sum().

Step 7: Divide the total marks by the number of subjects.

Step 8: Display the Average Marks.

Step 9: Stop.

## Algorithm 6: Find Highest Scorer

Step 1: Start.

Step 2: Check whether the student collection is empty.

Step 3: If empty, display "No Student Records Available".

Step 4: Otherwise, read each student's marks.

Step 5: Calculate the average marks of each student.

Step 6: Compare the averages and keep track of the highest average.

Step 7: Identify the student with the highest average.

Step 8: Display the Highest Scorer details.

Step 9: Stop.

## Algorithm 7: List Students by Department

Step 1: Start.

Step 2: Read the Department Name from the user.

Step 3: Search through all student records.

Step 4: Compare each student's department with the entered department.

Step 5: If a match is found, collect the matching student records.

Step 6: Display all matching students.

Step 7: If no match is found, display "No Students Found in This Department".

Step 8: Stop.

## Algorithm 8: Count Students

Step 1: Start.

Step 2: Access the student records collection.

Step 3: Count the total number of student records using len().

Step 4: Display the total number of students.

Step 5: Stop.

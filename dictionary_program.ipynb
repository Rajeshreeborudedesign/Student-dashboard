```python
# Dictionary Programs - Student Management System

# 1. Create Dictionary
student = {
    "roll": 101,
    "name": "Rajeshree",
    "department": "Computer",
    "marks": 90
}

print("Original Dictionary:", student)

# 2. Access Values
print("Student Name:", student["name"])

# 3. Keys Method
print("Keys:", student.keys())

# 4. Values Method
print("Values:", student.values())

# 5. Items Method
print("Items:", student.items())

# 6. Get Method
print("Department:", student.get("department"))

# 7. Update Method
student.update({"marks": 95})
print("After Update:", student)

# 8. Add New Key
student["email"] = "student@gmail.com"
print("After Adding Email:", student)

# 9. Pop Method
student.pop("email")
print("After Pop:", student)

# 10. Popitem Method
student.popitem()
print("After Popitem:", student)

# 11. Dictionary Traversal
print("Student Details:")

for key, value in student.items():
    print(key, ":", value)

# 12. Clear Method
student.clear()
print("After Clear:", student)
```

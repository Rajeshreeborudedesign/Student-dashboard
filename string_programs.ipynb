```python
# String Programs for Student Management System

student_name = "  rajeshree borude  "

print("Original Name:", student_name)

# 1. Remove extra spaces
print("Strip:", student_name.strip())

# 2. Uppercase
print("Uppercase:", student_name.upper())

# 3. Lowercase
print("Lowercase:", student_name.lower())

# 4. Title case
print("Title:", student_name.title())

# 5. Capitalize
print("Capitalize:", student_name.capitalize())

# 6. Length
print("Length:", len(student_name))

# 7. Slicing
print("First 5 characters:", student_name[2:7])

# 8. Replace
print("Replace:", student_name.replace("rajeshree", "student"))

# 9. Find
print("Find borude:", student_name.find("borude"))

# 10. Count
print("Count r:", student_name.count("r"))

# 11. Startswith
print("Starts with spaces:", student_name.startswith("  "))

# 12. Endswith
print("Ends with spaces:", student_name.endswith("  "))

# 13. Concatenation
department = "Computer"
print("Department:", student_name.strip() + " - " + department)

# 14. Split
print("Split:", student_name.split())

# 15. Join
subjects = ["Python", "Maths", "English"]
print("Subjects:", ", ".join(subjects))
```

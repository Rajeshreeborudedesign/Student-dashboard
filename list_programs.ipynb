```python
# List Programs - Student Management System

# 1. Create a List
students = ["Rajeshree", "Vivek", "Sneha", "Amit"]

print("Original List:", students)

# 2. Indexing
print("First Student:", students[0])

# 3. Traversal
print("All Students:")
for name in students:
    print(name)

# 4. Append
students.append("Priya")
print("After Append:", students)

# 5. Insert
students.insert(1, "Rahul")
print("After Insert:", students)

# 6. Extend
students.extend(["Neha", "Rohan"])
print("After Extend:", students)

# 7. Remove
students.remove("Amit")
print("After Remove:", students)

# 8. Pop
students.pop()
print("After Pop:", students)

# 9. Index
print("Index of Sneha:", students.index("Sneha"))

# 10. Count
print("Count of Rajeshree:", students.count("Rajeshree"))

# 11. Sort
students.sort()
print("Sorted List:", students)

# 12. Reverse
students.reverse()
print("Reversed List:", students)

# 13. List Comprehension
marks = [75, 85, 90, 65, 95]

high_marks = [m for m in marks if m >= 80]

print("Marks:", marks)
print("Marks above or equal to 80:", high_marks)

# 14. Clear
students.clear()
print("After Clear:", students)
```

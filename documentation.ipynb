# Student Record & Academic Management System

## Python Unit 2 Assignment Documentation

### 1. Introduction

The Student Record & Academic Management System is a console-based Python application designed to manage student information.

The system allows users to add, search, update, delete, and display student records. It also provides academic operations such as calculating average marks, finding the highest scorer, listing students by department, counting students, and generating reports.

### 2. Objectives

- To understand and implement Python data structures.
- To perform basic student record management operations.
- To calculate and analyze student marks.
- To implement a menu-driven Python application.
- To test the application using different test cases.

### 3. Technologies Used

- Programming Language: Python
- Development Environment: Jupyter Notebook
- Data Structures: String, List, Tuple, Set, Dictionary
- Interface: Console-based menu-driven system

### 4. Problem Analysis

The system is designed to solve the following problems:

1. Adding new student records.
2. Searching for a student using roll number.
3. Updating existing student information.
4. Deleting student records.
5. Displaying all student records.
6. Calculating average marks.
7. Finding the highest scorer.
8. Generating student reports.

### 5. Data Structure Selection

| Data Structure | Purpose |
|---|---|
| String | Stores student names and department names |
| List | Stores multiple student marks |
| Tuple | Represents fixed data |
| Set | Stores unique department names |
| Dictionary | Stores student records using roll numbers as keys |

### 6. System Features

The system provides a menu-driven interface through which users can perform student management and academic operations.

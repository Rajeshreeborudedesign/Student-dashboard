# STUDENT RECORD MANAGEMENT SYSTEM

## Pseudocode 1: Add Student

START

READ roll number

IF roll number already exists THEN
    DISPLAY "Duplicate Roll Number"
ELSE
    READ student details
    CREATE new student record
    STORE record in student collection
    DISPLAY "Student Added Successfully"
END IF

STOP

## Pseudocode 2: Search Student

START

READ roll number

SET found = FALSE

FOR EACH student IN student collection
    IF student roll number matches THEN
        DISPLAY student details
        SET found = TRUE
        EXIT LOOP
    END IF
END FOR

IF found = FALSE THEN
    DISPLAY "Student Not Found"
END IF

STOP

## Pseudocode 3: Update Student

START

READ roll number

IF roll number exists THEN
    DISPLAY update options
    READ selected field
    READ new value
    UPDATE selected field in student record
    DISPLAY "Student Updated Successfully"
ELSE
    DISPLAY "Student Not Found"
END IF

STOP

## Pseudocode 4: Delete Student

START

READ roll number

IF roll number exists THEN
    DELETE student record from collection
    DISPLAY "Student Deleted Successfully"
ELSE
    DISPLAY "Student Not Found"
END IF

STOP

## Pseudocode 5: Calculate Average

START

READ roll number

IF roll number exists THEN
    READ student marks

    IF marks are available THEN
        SET total = SUM of marks
        SET count = NUMBER of subjects
        SET average = total / count

        DISPLAY average marks
    ELSE
        DISPLAY "No Marks Available"
    END IF
ELSE
    DISPLAY "Student Not Found"
END IF

STOP

## Pseudocode 6: Find Highest Scorer

START

IF student collection is empty THEN
    DISPLAY "No Student Records Available"
ELSE
    SET highest_average = -1
    SET highest_student = NONE

    FOR EACH student IN student collection
        READ student marks

        IF marks are available THEN
            CALCULATE average marks

            IF average > highest_average THEN
                SET highest_average = average
                SET highest_student = student
            END IF
        END IF
    END FOR

## Pseudocode 7: List Students by Department

START

READ department name

SET found = FALSE

FOR EACH student IN student collection

    IF student's department matches entered department THEN
        DISPLAY student details
        SET found = TRUE
    END IF

END FOR

IF found = FALSE THEN
    DISPLAY "No Students Found in This Department"
END IF

STOP

## Pseudocode 8: Count Students

START

SET total = NUMBER OF RECORDS in student collection

DISPLAY "Total Students:", total

STOP

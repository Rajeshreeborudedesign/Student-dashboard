```python
# Set Programs - Student Management System

# 1. Create Set
departments = {"Computer", "Science", "Commerce"}

print("Original Set:", departments)

# 2. Add
departments.add("Arts")
print("After Add:", departments)

# 3. Remove
departments.remove("Science")
print("After Remove:", departments)

# 4. Discard
departments.discard("Commerce")
print("After Discard:", departments)

# 5. Membership
print("Computer Exists:", "Computer" in departments)

# 6. Union
set1 = {"Python", "Maths", "English"}
set2 = {"Python", "Physics", "Chemistry"}

print("Union:", set1.union(set2))

# 7. Intersection
print("Intersection:", set1.intersection(set2))

# 8. Difference
print("Difference:", set1.difference(set2))

# 9. Duplicate Removal
student_clubs = [
    "NSS", "Sports", "NSS",
    "Cultural", "Sports"
]

unique_clubs = set(student_clubs)

print("Unique Student Clubs:", unique_clubs)

# 10. Set Length
print("Total Unique Clubs:", len(unique_clubs))
```

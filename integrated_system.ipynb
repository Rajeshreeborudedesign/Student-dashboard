```python
# STUDENT RECORD & ACADEMIC MANAGEMENT SYSTEM

# Main collection to store student records
students = {}

# Set to store unique departments
departments = set()


# 1. ADD STUDENT
def add_student():
    roll = input("Enter Roll Number: ").strip()

    if not roll:
        print("Roll number cannot be empty!")
        return

    if roll in students:
        print("Duplicate Roll Number! Student already exists.")
        return

    name = input("Enter Student Name: ").strip()
    department = input("Enter Department: ").strip()
    email = input("Enter Email: ").strip()
    phone = input("Enter Phone Number: ").strip()
    address = input("Enter Address: ").strip()
    dob = input("Enter Date of Birth (DD-MM-YYYY): ").strip()
    registration = input("Enter Registration Number: ").strip()

    if not name or not department:
        print("Name and Department cannot be empty!")
        return

    subjects = []
    marks = []

    try:
        count = int(input("Enter Number of Subjects: "))

        if count <= 0:
            print("At least one subject is required!")
            return

        for i in range(count):
            subject = input(
                f"Enter Subject {i + 1}: "
            ).strip()

            if not subject:
                print("Subject name cannot be empty!")
                return

            mark = float(input(f"Enter Marks for {subject}: "))

            if mark < 0 or mark > 100:
                print("Marks must be between 0 and 100!")
                return

            subjects.append(subject)
            marks.append(mark)

        attendance = float(
            input("Enter Attendance Percentage (0-100): ")
        )

        if attendance < 0 or attendance > 100:
            print("Invalid Attendance!")
            return

    except ValueError:
        print("Invalid input! Enter numbers where required.")
        return

    # Tuple: fixed student identification details
    identity = (roll, registration, dob)

    # Dictionary: complete student profile
    students[roll] = {
        "identity": identity,
        "name": name,
        "department": department,
        "subjects": subjects,
        "marks": marks,
        "attendance": attendance,
        "contact": {
            "email": email,
            "phone": phone,
            "address": address
        }
    }

    departments.add(department)

    print("Student Added Successfully!")


# DISPLAY ONE STUDENT
def show_student(roll, student):
    print("\n---------------------------")
    print("Roll Number:", roll)
    print("Registration Number:", student["identity"][1])
    print("Date of Birth:", student["identity"][2])
    print("Name:", student["name"])
    print("Department:", student["department"])
    print("Subjects:", student["subjects"])
    print("Marks:", student["marks"])
    print("Attendance:", student["attendance"], "%")
    print("Contact:", student["contact"])

    if student["marks"]:
        average = sum(student["marks"]) / len(student["marks"])
        print("Average Marks:", round(average, 2))

    print("---------------------------")


# 2. SEARCH STUDENT
def search_student():
    roll = input("Enter Roll Number to Search: ").strip()

    if roll in students:
        show_student(roll, students[roll])
    else:
        print("Student Not Found!")


# 3. UPDATE STUDENT
def update_student():
    roll = input("Enter Roll Number to Update: ").strip()

    if roll not in students:
        print("Student Not Found!")
        return

    student = students[roll]

    print("\n1. Update Name")
    print("2. Update Department")
    print("3. Update Attendance")
    print("4. Update Email")
    print("5. Update Phone")
    print("6. Update Address")
    print("7. Update Subject Marks")

    choice = input("Enter Choice: ")

    if choice == "1":
        value = input("Enter New Name: ").strip()
        if not value:
            print("Name cannot be empty!")
            return
        student["name"] = value

    elif choice == "2":
        value = input("Enter New Department: ").strip()
        if not value:
            print("Department cannot be empty!")
            return
        student["department"] = value
        departments.clear()
        departments.update(
            s["department"] for s in students.values()
        )

    elif choice == "3":
        try:
            value = float(input("Enter New Attendance: "))
            if not 0 <= value <= 100:
                print("Invalid Attendance!")
                return
            student["attendance"] = value
        except ValueError:
            print("Enter a valid number!")
            return

    elif choice == "4":
        student["contact"]["email"] = input("New Email: ").strip()

    elif choice == "5":
        student["contact"]["phone"] = input("New Phone: ").strip()

    elif choice == "6":
        student["contact"]["address"] = input("New Address: ").strip()

    elif choice == "7":
        for i, subject in enumerate(student["subjects"]):
            print(i + 1, subject, student["marks"][i])

        try:
            number = int(input("Choose Subject Number: "))

            if number < 1 or number > len(student["subjects"]):
                print("Invalid Subject Number!")
                return

            mark = float(input("Enter New Marks: "))

            if not 0 <= mark <= 100:
                print("Marks must be between 0 and 100!")
                return

            student["marks"][number - 1] = mark

        except ValueError:
            print("Invalid Number!")
            return

    else:
        print("Invalid Choice!")
        return

    print("Student Updated Successfully!")


# 4. DELETE STUDENT
def delete_student():
    roll = input("Enter Roll Number to Delete: ").strip()

    if roll in students:
        del students[roll]

        departments.clear()
        departments.update(
            s["department"] for s in students.values()
        )

        print("Student Deleted Successfully!")
    else:
        print("Student Not Found!")


# 5. DISPLAY ALL RECORDS
def display_records():
    if not students:
        print("No Student Records Available!")
        return

    for roll, student in students.items():
        show_student(roll, student)


# 6. CALCULATE AVERAGE
def calculate_average():
    roll = input("Enter Roll Number: ").strip()

    if roll not in students:
        print("Student Not Found!")
        return

    marks = students[roll]["marks"]

    if not marks:
        print("No Marks Available!")
        return

    average = sum(marks) / len(marks)

    print("Average Marks:", round(average, 2))


# 7. FIND HIGHEST SCORER
def highest_scorer():
    if not students:
        print("No Student Records Available!")
        return

    highest_roll = None
    highest_average = -1

    for roll, student in students.items():
        marks = student["marks"]

        if marks:
            average = sum(marks) / len(marks)

            if average > highest_average:
                highest_average = average
                highest_roll = roll

    if highest_roll is None:
        print("No Marks Available!")
        return

    print("\nHighest Scorer:")
    show_student(highest_roll, students[highest_roll])


# 8. LIST STUDENTS BY DEPARTMENT
def by_department():
    department = input("Enter Department: ").strip()

    found = False

    for roll, student in students.items():
        if student["department"].lower() == department.lower():
            show_student(roll, student)
            found = True

    if not found:
        print("No Students Found in This Department!")


# 9. COUNT STUDENTS
def count_students():
    print("Total Students:", len(students))


# 10. GENERATE REPORTS
def generate_reports():
    if not students:
        print("No Records Available for Report!")
        return

    unique_subjects = set()
    all_marks = []

    for student in students.values():
        unique_subjects.update(student["subjects"])
        all_marks.extend(student["marks"])

    print("\n===== STUDENT REPORT =====")
    print("Total Students:", len(students))
    print("Unique Departments:", departments)
    print("Unique Subjects:", unique_subjects)

    if all_marks:
        overall_average = sum(all_marks) / len(all_marks)
        print("Overall Marks Average:", round(overall_average, 2))

    highest_scorer_report()


def highest_scorer_report():
    best_roll = None
    best_average = -1

    for roll, student in students.items():
        marks = student["marks"]

        if marks:
            average = sum(marks) / len(marks)

            if average > best_average:
                best_average = average
                best_roll = roll

    if best_roll is not None:
        print("Highest Scorer:", students[best_roll]["name"])
        print("Highest Average:", round(best_average, 2))


# MAIN MENU
while True:

    print("\n=================================")
    print(" STUDENT RECORD MANAGEMENT SYSTEM")
    print("=================================")

    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display Records")
    print("6. Calculate Average")
    print("7. Highest Scorer")
    print("8. By Department")
    print("9. Count Students")
    print("10. Generate Reports")
    print("0. Exit")

    choice = input("\nEnter Your Choice: ").strip()

    if choice == "1":
        add_student()

    elif choice == "2":
        search_student()

    elif choice == "3":
        update_student()

    elif choice == "4":
        delete_student()

    elif choice == "5":
        display_records()

    elif choice == "6":
        calculate_average()

    elif choice == "7":
        highest_scorer()

    elif choice == "8":
        by_department()

    elif choice == "9":
        count_students()

    elif choice == "10":
        generate_reports()

    elif choice == "0":
        print("Thank You! Program Closed.")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```
